# **Preprocessing Store Data**

### **Data Loading**

```python
logger.info("Loading Store Dataset")
df_store = pd.read_csv("store.csv")
```

This cell loads the `store.csv` dataset into a pandas DataFrame called `df_store`. This dataset contains information about the individual stores.

```python
df_store.head()
```

This command displays the first 5 rows of the `df_store` DataFrame, providing a quick preview of the data structure and content.

```python
df_store.tail()
```

This command displays the last 5 rows of the `df_store` DataFrame, useful for checking the end of the dataset and ensuring data consistency.

```python
df_store.info()
```

This cell provides a concise summary of the `df_store` DataFrame, including the column names, non-null counts, data types, and memory usage. It's crucial for understanding the data's integrity and identifying potential issues like missing values.

### **Check Missing Value**

```python
df_store.isnull().sum()
```

This cell calculates and displays the total number of missing values for each column in the `df_store` DataFrame. This helps in identifying which columns require missing value imputation or handling.

### **Handle Missing Value**

```python
df_store['CompetitionDistance'] = df_store['CompetitionDistance'].fillna(df_store['CompetitionDistance'].median())

df_store['CompetitionOpenSinceMonth'] = df_store['CompetitionOpenSinceMonth'].fillna(df_store['CompetitionOpenSinceMonth'].median())

df_store['CompetitionOpenSinceYear'] = df_store['CompetitionOpenSinceYear'].fillna(df_store['CompetitionOpenSinceYear'].median())

df_store['PromoInterval'] = df_store['PromoInterval'].fillna(df_store['PromoInterval'].mode())
df_store['Promo2SinceWeek'] = df_store['Promo2SinceWeek'].fillna(df_store['Promo2SinceWeek'].median())
df_store['Promo2SinceYear'] = df_store['Promo2SinceYear'].fillna(df_store['Promo2SinceYear'].median())

```

This cell handles missing values in `df_store` by imputing them. `CompetitionDistance`, `CompetitionOpenSinceMonth`, `CompetitionOpenSinceYear`, `Promo2SinceWeek`, and `Promo2SinceYear` are filled with their respective medians, while `PromoInterval` is filled with its mode. This ensures that all columns have complete data for further analysis.

### **Fixing Data Type**

```python
df_store['Promo2SinceWeek'] = df_store['Promo2SinceWeek'].astype(int)
df_store['Promo2SinceYear'] = df_store['Promo2SinceYear'].astype(int)
df_store['CompetitionOpenSinceMonth'] = df_store['CompetitionOpenSinceMonth'].astype(int)
df_store['CompetitionOpenSinceYear'] = df_store['CompetitionOpenSinceYear'].astype(int)
```

This cell converts the data types of `Promo2SinceWeek`, `Promo2SinceYear`, `CompetitionOpenSinceMonth`, and `CompetitionOpenSinceYear` columns to integer type. This is done after filling missing values to ensure consistency and appropriate data representation.

### **Check Duplicate Value**

```python
df_store.duplicated().sum()
```

This cell checks for any duplicate rows in the `df_store` DataFrame. The output indicates that there are no duplicate entries, ensuring data uniqueness.

```python
logger.info("Preprocessing Store Dataset Completed")
```

This line logs a message confirming that the preprocessing steps for the `store` dataset have been successfully completed.

# **Preprocessing Test Data**

### **Data Loading**

```python
logger.info("Loading Test Dataset")
df_test = pd.read_csv("test.csv")
```

This cell loads the `test.csv` dataset into a pandas DataFrame called `df_test`. This dataset contains the data for which sales predictions need to be made.

```python
df_test.head()
```

This command displays the first 5 rows of the `df_test` DataFrame, providing a quick preview of the test data structure and content.

```python
df_test.tail()
```

This command displays the last 5 rows of the `df_test` DataFrame, useful for checking the end of the dataset and ensuring data consistency.

```python
df_test.info()
```

This cell provides a concise summary of the `df_test` DataFrame, including the column names, non-null counts, data types, and memory usage. It's crucial for understanding the data's integrity and identifying potential issues like missing values.

### **Check Missing Value**

```python
df_test.isnull().sum()
```

This cell calculates and displays the total number of missing values for each column in the `df_test` DataFrame. This helps in identifying which columns require missing value imputation or handling.

### **Handle Missing Value**

```python
df_test['Open'] = df_test['Open'].fillna(df_test['DayOfWeek'].apply(lambda x: 0 if x == 7 else 1))
```

This cell handles the missing values in the 'Open' column of `df_test`. It imputes 'Open' values based on the 'DayOfWeek' column: if the `DayOfWeek` is 7 (Sunday), 'Open' is set to 0 (closed); otherwise, it's set to 1 (open).

### **Fixing Data Type**

```python
df_test['Open'] = df_test['Open'].astype(int)
df_test['Date'] = pd.to_datetime(df_test['Date'])
df_test['StateHoliday'] = df_test['StateHoliday'].astype(str)
```

This cell converts the 'Open' column to an integer type, the 'Date' column to a datetime object for time-series analysis, and ensures the 'StateHoliday' column is treated as a string in `df_test`.

### **Check Duplicate Value**

```python
df_test.duplicated().sum()
```

This cell checks for any duplicate rows in the `df_test` DataFrame. The output indicates that there are no duplicate entries, ensuring data uniqueness.

# **Merge Store & Train Dataset**

```python
logger.info("Merging Store and Train Dataset")
store_train = pd.merge(df_train, df_store, on='Store', how='left')
print("Train Merged Shape:",store_train.shape)
```

This cell merges the `df_train` and `df_store` DataFrames into a new DataFrame called `store_train`. The merge is performed on the 'Store' column using a left join, combining store-specific information with the training sales data. It also prints the shape of the merged DataFrame.

# **Merge Store & Test Dataset**

```python
logger.info("Merging Store and Test Dataset")
store_test = pd.merge(df_test, df_store, on='Store', how='left')
print("Test Merged Shape:",store_test.shape)
```

This cell merges the `df_test` and `df_store` DataFrames into a new DataFrame called `store_test`. Similar to the training data, this merges store-specific information with the test sales data for prediction. It also prints the shape of the merged DataFrame.

# **Saving Merged Dataset to CSV**

```python
store_train.to_csv('store_train.csv', index=False)
store_test.to_csv('store_test.csv', index=False)
logger.info("Saving store_train & store_test dataset to CSV Successfully")
```

This cell saves the newly created `store_train` and `store_test` DataFrames to CSV files named 'store_train.csv' and 'store_test.csv' respectively. The `index=False` argument prevents pandas from writing the DataFrame index as a column in the CSV file. A log message confirms the successful saving.
